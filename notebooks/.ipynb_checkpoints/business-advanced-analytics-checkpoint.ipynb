```python
#vamos a por el mes de la sesion, la semana
visitas['sesion_mes'] = pd.to_datetime(visitas['start_ts'].dt.strftime('%Y-%m-01'))
visitas['sesion_semana'] = visitas['start_ts'].dt.isocalendar().week
visitas['sesion_fecha'] = visitas['start_ts'].dt.date


# string f time es para formatear fechas y extraer lo que necesitamos de la fecha
```

```python
visitas.head()
```

```python
#el mesno lleva consigo el año en el que pertenece
```

```python
# ahora podemos sacar DAU, WAU Y MAU
# DAILY AVERAGE USERS, WEEKLY Y MONTHLY
```

```python
dau = visitas.groupby('sesion_fecha').agg({'uid':'nunique'})

wau = visitas.groupby('sesion_semana').agg({'uid':'nunique'})

mau = visitas.groupby('sesion_mes').agg({'uid':'nunique'})
```

```python
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots(1, 3, figsize=(19,5))
ax[0].plot(dau)
ax[0].set(title='DAU',xlabel='fecha',ylabel='visitantes')
ax[1].plot(wau)
ax[1].set(title='WAU',xlabel='fecha',ylabel='visitantes')
ax[2].plot(mau)
ax[2].set(title='MAU',xlabel='fecha',ylabel='visitantes')
fig.autofmt_xdate(rotation=30)
plt.show()
```

```python
primeras_visitas = (
    visitas.sort_values('sesion_fecha')
           .groupby('uid', as_index=False)
           .first()[['uid', 'sesion_mes', 'sesion_semana', 'sesion_fecha']]
           .rename(columns={
               'sesion_mes': 'primera_sesion_mes',
               'sesion_semana': 'primera_sesion_semana',
               'sesion_fecha': 'primera_sesion_fecha'
           })
)
```

```python
#si no haces que se distinga el mes y el año o semana por año, vas a causar desastre.
```

```python
visitas_full = pd.merge(primeras_visitas,visitas,on='uid',how='inner')
```

```python
import numpy as np
visitas_full['edad_meses']=((visitas_full['sesion_mes']-visitas_full['primera_sesion_mes'])/(30*np.timedelta64(1,'D'))).round().astype('int')
#el time delta es el cambio. divido para 30 pq la resta me sale en dias
```

```python
#pivoteamos la tabla para hacer una visita cohorte
cohortes_visitas = visitas_full.pivot_table(
    index='primera_sesion_mes',
    columns = 'edad_meses',
    values = 'uid',
    aggfunc='nunique'
)
```

```python
cohortes_visitas
```
